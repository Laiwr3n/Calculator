```python

import random
import string
import os

# --- 工具类 ---
def bmi_calculator():
    print("\n 欢迎使用 BMI 计算器")
    try:
        height = float(input("请输入你的身高（单位：米）："))
        weight = float(input("请输入你的体重（单位：公斤）："))
        bmi = weight / (height ** 2)
        print(f"你的 BMI 是：{bmi:.2f}")
        if bmi < 18.5:
            print(" 体重过轻")
        elif bmi < 24.9:
            print(" 正常范围")
        elif bmi < 29.9:
            print(" 体重偏重")
        else:
            print(" 肥胖")
    except ValueError:
        print(" 输入无效，请输入数字！")

def calculator():
    print("\n 欢迎使用简单计算器！")
    print("你可以输入加（+）、减（-）、乘（*）、除（/）")
    while True:
        try:
            expression = input("请输入一个数学表达式（或输入 q 退出）：")
            if expression.lower() == 'q':
                break
            result = eval(expression)
            print(f"计算结果：{result}")
        except Exception as e:
            print(f" 错误：{e}")

def unit_converter():
    print("\n 欢迎使用单位换算器！")
    print("你可以选择进行以下换算：")
    print("1. 长度换算")
    print("2. 体重换算")
    print("3. 温度换算")

    choice = input("请选择你要进行的换算（1/2/3）：")

    if choice == "1":
        print("\n请选择单位：")
        print("1. 米 -> 千米")
        print("2. 千米 -> 米")
        length_choice = input("请输入选项（1/2）：")
        if length_choice == "1":
            value = float(input("请输入米数："))
            print(f"{value} 米 = {value / 1000} 千米")
        elif length_choice == "2":
            value = float(input("请输入千米数："))
            print(f"{value} 千米 = {value * 1000} 米")
        else:
            print("无效选项！")

    elif choice == "2":
        print("\n请选择单位：")
        print("1. 公斤 -> 克")
        print("2. 克 -> 公斤")
        weight_choice = input("请输入选项（1/2）：")
        if weight_choice == "1":
            value = float(input("请输入公斤数："))
            print(f"{value} 公斤 = {value * 1000} 克")
        elif weight_choice == "2":
            value = float(input("请输入克数："))
            print(f"{value} 克 = {value / 1000} 公斤")
        else:
            print("无效选项！")

    elif choice == "3":
        print("\n请选择单位：")
        print("1. 摄氏度 -> 华氏度")
        print("2. 华氏度 -> 摄氏度")
        temperature_choice = input("请输入选项（1/2）：")
        if temperature_choice == "1":
            value = float(input("请输入摄氏度："))
            print(f"{value} 摄氏度 = {(value * 9/5) + 32} 华氏度")
        elif temperature_choice == "2":
            value = float(input("请输入华氏度："))
            print(f"{value} 华氏度 = {(value - 32) * 5/9} 摄氏度")
        else:
            print("无效选项！")
    else:
        print("无效选择！")

def password_generator():
    print("\n欢迎使用随机密码生成器！")
    length = int(input("请输入密码的长度："))
    include_special = input("是否包含特殊字符（y/n）：").lower() == 'y'

    all_characters = string.ascii_letters + string.digits
    if include_special:
        all_characters += string.punctuation

    password = ''.join(random.choice(all_characters) for _ in range(length))
    print(f"生成的随机密码是：{password}")

def clear_screen():
    os.system('cls' if os.name == 'nt' else 'clear')

def fruit_summary():
    fruits = {}

    while True:
        fruit_type = input("请输入水果种类（输入 'n' 结束）：").strip()
        if fruit_type.lower() == 'n':
            break

        try:
            cost = float(input(f"请输入 {fruit_type} 的单价："))
            quantity = int(input(f"请输入 {fruit_type} 的数量："))
        except ValueError:
            print("无效输入，请输入数字！")
            continue

        if fruit_type in fruits:
            fruits[fruit_type]["quantity"] += quantity
            fruits[fruit_type]["total_cost"] += cost * quantity
        else:
            fruits[fruit_type] = {
                "quantity": quantity,
                "total_cost": cost * quantity,
                "unit_price": cost
            }

    fruit_types = len(fruits)
    total_quantity = sum(f["quantity"] for f in fruits.values())
    total_cost = sum(f["total_cost"] for f in fruits.values())
    average_price = total_cost / total_quantity if total_quantity > 0 else 0
    most_expensive = max(fruits.items(), key=lambda x: x[1]["unit_price"]) if fruits else (None, {"unit_price": 0})

    print("\n水果统计结果")
    print(f"水果种类数：{fruit_types}")
    print(f"水果总数量：{total_quantity}")
    print(f"水果总花费：${total_cost:.2f}")
    print(f"平均单价：${average_price:.2f}")
    print(f"最贵的水果是：{most_expensive[0]}，单价为 ${most_expensive[1]['unit_price']:.2f}")

# --- 游戏类 ---
def number_guess_game():
    print("\n欢迎来到数字猜猜猜！")
    print("1. 轻松模式（无限机会）")
    print("2. 挑战模式（10次机会）")
    print("3. 创意模式（自定义范围+次数）")
    mode = input("请输入模式编号 (1/2/3): ")

    if mode == "1":
        min_num, max_num, chances = 1, 100, None
    elif mode == "2":
        min_num, max_num, chances = 1, 100, 10
    elif mode == "3":
        min_num = int(input("请输入最小值："))
        max_num = int(input("请输入最大值："))
        chances = int(input("请输入最大猜测次数："))
    else:
        print("输入无效，返回主菜单～")
        return

    answer = random.randint(min_num, max_num)
    tries = 0
    print(f"游戏开始！猜一个 {min_num} 到 {max_num} 之间的数字吧！")

    while True:
        try:
            guess = int(input("请输入你的猜测："))
        except ValueError:
            print("请输入一个数字！")
            continue

        tries += 1

        if guess < answer:
            print("太小了！")
        elif guess > answer:
            print("太大了！")
        else:
            print(f"恭喜你猜对啦！正确答案是 {answer}，你用了 {tries} 次机会！")
            break

        if chances and tries >= chances:
            print(f"😢 游戏结束！正确答案是 {answer}。")
            break

def rock_paper_scissors():
    print("\n欢迎来到石头剪刀布大对决！")
    choices = ['石头', '剪刀', '布']
    win_map = {'石头': '剪刀', '剪刀': '布', '布': '石头'}
    number_map = {'1': '石头', '2': '剪刀', '3': '布'}

    while True:
        print("\n1. 石头 \n2. 剪刀 \n3. 布 \nq. 退出游戏")
        player = input("你选的是：").strip()
        if player.lower() == 'q':
            print("回到主菜单！")
            break
        if player in number_map:
            player = number_map[player]
        if player not in choices:
            print("无效选择")
            continue

        computer = random.choice(choices)
        print(f"电脑出了：{computer}")

        if player == computer:
            print("平局！")
        elif win_map[player] == computer:
            print("你赢了！")
        else:
            print("你输了！")

def math_quiz():
    print("\n欢迎来到简易数学小测验！")
    score = 0
    for i in range(5):
        a = random.randint(1, 20)
        b = random.randint(1, 20)
        op = random.choice(['+', '-', '*'])
        answer = eval(f"{a}{op}{b}")
        try:
            guess = int(input(f"第 {i+1} 题：{a} {op} {b} = "))
        except ValueError:
            print("输入无效")
            continue
        if guess == answer:
            print("正确！")
            score += 1
        else:
            print(f"错啦，正确答案是 {answer}")
    print(f"📊 得分：{score}/5")

def number_fill_in_the_blank():
    print("\n数字填空游戏！")
    a, b, c = random.randint(1, 20), random.randint(1, 20), random.randint(1, 20)
    print(f"填空：{a} + ? = {b} * {c}")
    answer = b * c - a
    try:
        guess = int(input("你的答案："))
        if guess == answer:
            print("🎉 正确！")
        else:
            print(f"错误，正确答案是：{answer}")
    except ValueError:
        print("请输入数字！")

def whack_a_mole():
    print("\n打地鼠游戏！")
    score = 0
    for _ in range(5):
        pos = random.randint(1, 5)
        print(f"\n地鼠出现在位置 {pos}！")
        try:
            guess = int(input("你打哪一个位置？(1-5)："))
            if guess == pos:
                print("打中！")
                score += 1
            else:
                print("没中！")
        except ValueError:
            print("无效输入")
    print(f"游戏结束，你打中了 {score} 次！")

def guess_riddle():
    print("\n猜谜语游戏！")
    riddles = [
        {"riddle": "什么东西越洗越脏？", "answer": "水"},
        {"riddle": "我有四条腿，但不能走路，我是什么？", "answer": "桌子"},
        {"riddle": "什么字永远写不完？", "answer": "田"}
    ]
    r = random.choice(riddles)
    guess = input(f"谜语：{r['riddle']}\n你的答案是：").strip()
    if guess == r['answer']:
        print("正确！")
    else:
        print(f"错了！答案是：{r['answer']}")

# --- 菜单 ---
def game_menu():
    while True:
        print("\n小游戏菜单")
        print("1. 猜数字")
        print("2. 石头剪刀布")
        print("3. 数学小测验")
        print("4. 数字填空")
        print("5. 打地鼠")
        print("6. 猜谜语")
        print("0. 返回主菜单")
        choice = input("请选择：")
        if choice == "1":
            number_guess_game()
        elif choice == "2":
            rock_paper_scissors()
        elif choice == "3":
            math_quiz()
        elif choice == "4":
            number_fill_in_the_blank()
        elif choice == "5":
            whack_a_mole()
        elif choice == "6":
            guess_riddle()
        elif choice == "0":
            break
        else:
            print("请选择有效数字！")

def tool_menu():
    while True:
        print("\n🛠️ 工具菜单")
        print("1. BMI 计算器")
        print("2. 简易计算器")
        print("3. 单位换算器")
        print("4. 密码生成器")
        print("5. 水果统计器")
        print("0. 返回主菜单")
        choice = input("请选择：")
        if choice == "1":
            bmi_calculator()
        elif choice == "2":
            calculator()
        elif choice == "3":
            unit_converter()
        elif choice == "4":
            password_generator()
        elif choice == "5":
            fruit_summary()
        elif choice == "0":
            break
        else:
            print("请选择有效数字！")

# --- 主菜单 ---
def main_menu():
    while True:
        print("\n欢迎来到娱乐工具大合集！")
        print("1. 玩游戏")
        print("2. 用工具")
        print("0. 退出程序")
        main_choice = input("请选择：")
        if main_choice == "1":
            game_menu()
        elif main_choice == "2":
            clear_screen()
            tool_menu()
        elif main_choice == "0":
            print("再见啦！")
            break
        else:
            print("无效选择！")

# 入口
if __name__ == "__main__":
    main_menu()
```

```python
import random
import string
import os

# --- Utility Functions ---
def bmi_calculator():
    print("\n Welcome to the BMI Calculator")
    try:
        height = float(input("Enter your height (meters): "))
        weight = float(input("Enter your weight (kg): "))
        bmi = weight / (height ** 2)
        print(f"Your BMI is: {bmi:.2f}")
        if bmi < 18.5:
            print(" Underweight")
        elif bmi < 24.9:
            print(" Normal weight")
        elif bmi < 29.9:
            print(" Overweight")
        else:
            print(" Obese")
    except ValueError:
        print(" Invalid input, please enter numbers!")

def calculator():
    print("\n Welcome to the Simple Calculator!")
    print("You can use +, -, *, /")
    while True:
        try:
            expression = input("Enter a math expression (or 'q' to quit): ")
            if expression.lower() == 'q':
                break
            result = eval(expression)
            print(f"Result: {result}")
        except Exception as e:
            print(f" Error: {e}")

def unit_converter():
    print("\n Welcome to the Unit Converter!")
    print("You can choose from the following conversions:")
    print("1. Length Conversion")
    print("2. Weight Conversion")
    print("3. Temperature Conversion")

    choice = input("Choose a conversion (1/2/3): ")

    if choice == "1":
        print("\nChoose units:")
        print("1. Meters -> Kilometers")
        print("2. Kilometers -> Meters")
        length_choice = input("Enter option (1/2): ")
        if length_choice == "1":
            value = float(input("Enter meters: "))
            print(f"{value} meters = {value / 1000} kilometers")
        elif length_choice == "2":
            value = float(input("Enter kilometers: "))
            print(f"{value} kilometers = {value * 1000} meters")
        else:
            print("Invalid option!")

    elif choice == "2":
        print("\nChoose units:")
        print("1. Kilograms -> Grams")
        print("2. Grams -> Kilograms")
        weight_choice = input("Enter option (1/2): ")
        if weight_choice == "1":
            value = float(input("Enter kilograms: "))
            print(f"{value} kg = {value * 1000} g")
        elif weight_choice == "2":
            value = float(input("Enter grams: "))
            print(f"{value} g = {value / 1000} kg")
        else:
            print("Invalid option!")

    elif choice == "3":
        print("\nChoose units:")
        print("1. Celsius -> Fahrenheit")
        print("2. Fahrenheit -> Celsius")
        temperature_choice = input("Enter option (1/2): ")
        if temperature_choice == "1":
            value = float(input("Enter Celsius: "))
            print(f"{value} °C = {(value * 9/5) + 32} °F")
        elif temperature_choice == "2":
            value = float(input("Enter Fahrenheit: "))
            print(f"{value} °F = {(value - 32) * 5/9} °C")
        else:
            print("Invalid option!")
    else:
        print("Invalid selection!")

def password_generator():
    print("\nWelcome to the Random Password Generator!")
    length = int(input("Enter desired password length: "))
    include_special = input("Include special characters? (y/n): ").lower() == 'y'

    all_characters = string.ascii_letters + string.digits
    if include_special:
        all_characters += string.punctuation

    password = ''.join(random.choice(all_characters) for _ in range(length))
    print(f"Generated password: {password}")

def clear_screen():
    os.system('cls' if os.name == 'nt' else 'clear')

def fruit_summary():
    fruits = {}

    while True:
        fruit_type = input("Enter fruit type ('n' to finish): ").strip()
        if fruit_type.lower() == 'n':
            break

        try:
            cost = float(input(f"Enter unit price for {fruit_type}: "))
            quantity = int(input(f"Enter quantity of {fruit_type}: "))
        except ValueError:
            print("Invalid input, please enter numbers!")
            continue

        if fruit_type in fruits:
            fruits[fruit_type]["quantity"] += quantity
            fruits[fruit_type]["total_cost"] += cost * quantity
        else:
            fruits[fruit_type] = {
                "quantity": quantity,
                "total_cost": cost * quantity,
                "unit_price": cost
            }

    fruit_types = len(fruits)
    total_quantity = sum(f["quantity"] for f in fruits.values())
    total_cost = sum(f["total_cost"] for f in fruits.values())
    average_price = total_cost / total_quantity if total_quantity > 0 else 0
    most_expensive = max(fruits.items(), key=lambda x: x[1]["unit_price"]) if fruits else (None, {"unit_price": 0})

    print("\nFruit Summary")
    print(f"Types of fruits: {fruit_types}")
    print(f"Total quantity: {total_quantity}")
    print(f"Total cost: ${total_cost:.2f}")
    print(f"Average unit price: ${average_price:.2f}")
    print(f"Most expensive fruit: {most_expensive[0]} at ${most_expensive[1]['unit_price']:.2f}")

# --- Games ---
def number_guess_game():
    print("\nWelcome to Number Guessing Game!")
    print("1. Easy Mode (Unlimited tries)")
    print("2. Challenge Mode (10 tries)")
    print("3. Custom Mode (Set range & tries)")
    mode = input("Choose a mode (1/2/3): ")

    if mode == "1":
        min_num, max_num, chances = 1, 100, None
    elif mode == "2":
        min_num, max_num, chances = 1, 100, 10
    elif mode == "3":
        min_num = int(input("Enter minimum number: "))
        max_num = int(input("Enter maximum number: "))
        chances = int(input("Enter max number of attempts: "))
    else:
        print("Invalid input, returning to main menu.")
        return

    answer = random.randint(min_num, max_num)
    tries = 0
    print(f"Game start! Guess a number between {min_num} and {max_num}.")

    while True:
        try:
            guess = int(input("Your guess: "))
        except ValueError:
            print("Please enter a number!")
            continue

        tries += 1

        if guess < answer:
            print("Too low!")
        elif guess > answer:
            print("Too high!")
        else:
            print(f"🎉 Correct! The answer is {answer}. You used {tries} attempts.")
            break

        if chances and tries >= chances:
            print(f"😢 Game over! The correct answer was {answer}.")
            break

def rock_paper_scissors():
    print("\nWelcome to Rock Paper Scissors Showdown!")
    choices = ['Rock', 'Scissors', 'Paper']
    win_map = {'Rock': 'Scissors', 'Scissors': 'Paper', 'Paper': 'Rock'}
    number_map = {'1': 'Rock', '2': 'Scissors', '3': 'Paper'}

    while True:
        print("\n1. Rock \n2. Scissors \n3. Paper \nq. Quit")
        player = input("Your choice: ").strip()
        if player.lower() == 'q':
            print("Returning to main menu!")
            break
        if player in number_map:
            player = number_map[player]
        if player not in choices:
            print("Invalid choice")
            continue

        computer = random.choice(choices)
        print(f"Computer chose: {computer}")

        if player == computer:
            print("It's a draw!")
        elif win_map[player] == computer:
            print("You win!")
        else:
            print("You lose!")

def math_quiz():
    print("\nWelcome to Math Quiz!")
    score = 0
    for i in range(5):
        a = random.randint(1, 20)
        b = random.randint(1, 20)
        op = random.choice(['+', '-', '*'])
        answer = eval(f"{a}{op}{b}")
        try:
            guess = int(input(f"Q{i+1}: {a} {op} {b} = "))
        except ValueError:
            print("Invalid input")
            continue
        if guess == answer:
            print("Correct!")
            score += 1
        else:
            print(f"Wrong, the answer was {answer}")
    print(f"📊 Score: {score}/5")

def number_fill_in_the_blank():
    print("\nNumber Fill-in-the-Blank Game!")
    a, b, c = random.randint(1, 20), random.randint(1, 20), random.randint(1, 20)
    print(f"Fill in the blank: {a} + ? = {b} * {c}")
    answer = b * c - a
    try:
        guess = int(input("Your answer: "))
        if guess == answer:
            print("🎉 Correct!")
        else:
            print(f"Wrong, the answer was: {answer}")
    except ValueError:
        print("Please enter a number!")

def whack_a_mole():
    print("\nWhack-a-Mole Game!")
    score = 0
    for _ in range(5):
        pos = random.randint(1, 5)
        print(f"\nA mole appeared at position {pos}!")
        try:
            guess = int(input("Which position do you hit? (1-5): "))
            if guess == pos:
                print("Hit!")
                score += 1
            else:
                print("Missed!")
        except ValueError:
            print("Invalid input")
    print(f"Game over! You hit {score} moles!")

def guess_riddle():
    print("\nRiddle Game!")
    riddles = [
        {"riddle": "What gets dirtier the more you wash it?", "answer": "water"},
        {"riddle": "I have four legs but cannot walk, what am I?", "answer": "table"},
        {"riddle": "Which character can never be fully written?", "answer": "field"}
    ]
    r = random.choice(riddles)
    guess = input(f"Riddle: {r['riddle']}\nYour answer: ").strip()
    if guess == r['answer']:
        print("Correct!")
    else:
        print(f"Wrong! The answer is: {r['answer']}")

# --- Menus ---
def game_menu():
    while True:
        print("\n🎮 Game Menu")
        print("1. Number Guessing")
        print("2. Rock Paper Scissors")
        print("3. Math Quiz")
        print("4. Fill in the Number")
        print("5. Whack-a-Mole")
        print("6. Guess the Riddle")
        print("0. Return to Main Menu")
        choice = input("Choose an option: ")
        if choice == "1":
            number_guess_game()
        elif choice == "2":
            rock_paper_scissors()
        elif choice == "3":
            math_quiz()
        elif choice == "4":
            number_fill_in_the_blank()
        elif choice == "5":
            whack_a_mole()
        elif choice == "6":
            guess_riddle()
        elif choice == "0":
            break
        else:
            print("Please enter a valid number!")

def tool_menu():
    while True:
        print("\n🛠️ Tools Menu")
        print("1. BMI Calculator")
        print("2. Simple Calculator")
        print("3. Unit Converter")
        print("4. Password Generator")
        print("5. Fruit Summary Tool")
        print("0. Return to Main Menu")
        choice = input("Choose an option: ")
        if choice == "1":
            bmi_calculator()
        elif choice == "2":
            calculator()
        elif choice == "3":
            unit_converter()
        elif choice == "4":
            password_generator()
        elif choice == "5":
            fruit_summary()
        elif choice == "0":
            break
        else:
            print("Please enter a valid number!")

# --- Main Menu ---
def main_menu():
    while True:
        print("\n🎉 Welcome to the Ultimate Fun Tool Collection!")
        print("1. Play Games")
        print("2. Use Tools")
        print("0. Exit")
        main_choice = input("Choose an option: ")
        if main_choice == "1":
            game_menu()
        elif main_choice == "2":
            clear_screen()
            tool_menu()
        elif main_choice == "0":
            print("Goodbye!")
            break
        else:
            print("Invalid choice!")

# Entry Point
if __name__ == "__main__":
    main_menu()
```
